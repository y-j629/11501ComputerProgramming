使用者輸入：請輸入正整數 $n$，其中 $1\le n\le 9$，輸出 $n\times n$ 乘法表。格式以 $n=5$ 為例。

1. 只能使用 `for`。
2. 字串格式化只能使用 f-string 對齊。

```python
n = int(input("請輸入正整數 n（1 到 9）："))

if 1 <= n <= 9:
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            print(f"{i:>3}×{j:>3}={i * j:>3}", end="   ")
        print()
else:
    print("輸入錯誤，n 必須介於 1 到 9。")
```

使用者輸入：請輸入正整數 $n$，其中 $1\le n\le 9$，輸出二進位 $n\times n$ 乘法表。格式以 $n=5$ 為例，參考題目圖片。

1. 只能使用 `while`。
2. 字串格式化只能使用 f-string 對齊。

乘數以 3 位二進位表示，乘積以 6 位二進位表示，例如 `001×011=000011`。

```python
n = int(input("請輸入正整數 n（1 到 9）："))

if 1 <= n <= 9:
    i = 1
    while i <= n:
        j = 1
        while j <= n:
            print(f"{i:03b}×{j:03b}={i * j:06b}", end="   ")
            j += 1
        print()
        i += 1
else:
    print("輸入錯誤，n 必須介於 1 到 9。")
```
