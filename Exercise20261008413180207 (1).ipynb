1. 只能使用for。
3. 字串格式化只能使用 f-string 對齊。

```python
n = int(input("請輸入正整數 n（1 到 9）："))

if 1 <= n <= 9:
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            print(f"{i:>3}×{j:>3}={i * j:>3}", end="   ")
        print()
else:
    print("輸入錯誤，n 必須介於 1 到 9。")
```

以下為題目提供的二進位乘法表格式：



```python
n = int(input("請輸入正整數 n（1 到 9）："))

if 1 <= n <= 9:
    i = 1
    while i <= n:
        j = 1
        while j <= n:
            print(f"{i:03b}×{j:03b}={i * j:06b}", end="   ")
            j += 1
        print()
        i += 1
else:
    print("輸入錯誤，n 必須介於 1 到 9。")
```
